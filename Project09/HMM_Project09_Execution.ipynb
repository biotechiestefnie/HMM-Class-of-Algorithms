## Imports and Matrix Formatting Helper Function:
The following cell imports all packages and modules needed to call our script and supporting module for execution in this notebook. The helper function included directly below the imports is used only for formatting the output matrices produced during execution, strictly for visualization purposes in Jupyter Notebooks and in the GitHub repository for this project, HMM_Class_of_Algorithms.

```python
# Import packages and modules
import numpy as np
from project09 import HMMModel
from dict_maker import StrMatrix
from IPython.display import display, HTML
from tabulate import tabulate

def show_matrix_html(matrix, float_format="{:.3e}"):
    """
    Function to display matrices output from execution of project09.py in an organized, interpretable format.
    Parameters:
         matrix (np.array): output matrix from each HMMModel method execution
         float_format (str, optional): formatting string for floating point output
    """
    formatted = []
    for row in matrix:
        new_row = []
        for cell in row:
            if isinstance(cell, float):
                new_row.append(float_format.format(cell))
            else:
                new_row.append(str(cell))
        formatted.append(new_row)

    return HTML(tabulate(formatted, tablefmt="html"))


def add_headers(matrix, states, observation):
    """
    Add state labels and observation characters as headers for a DP matrix. This is for
    forward and backward algorithms to display headers and hidden states because they are
    numpy arrays.
    Steps:
         Create a header row containing a blank cell followed by each observed character.
        Create one row per hidden state, placing the state label in column 0.
        Append the corresponding matrix row values after the state label.
        Return the combined header row and state-labeled rows as a nested list.
    """
    header_row = [" "] + list(observation)
    rows = []
    for i, state in enumerate(states):
        rows.append([state] + list(matrix[i]))
    return [header_row] + rows
```

## Defining Probabilities
### Viterbi Probability Dictionaries

```python
# Setting Viterbi probability dicts
initial_probs_v = {
    "I": 0.2,
    "G": 0.8
}

transition_probs_v = {
    "I": {"I": 0.1, "G": 0.9},
    "G": {"I": 0.6, "G": 0.4}
}

emission_probs_v = {
    "I": {"A": 0.1, "T": 0.1, "C": 0.4, "G": 0.4},
    "G": {"A": 0.15, "T": 0.15, "C": 0.3, "G": 0.3}
}
```

### Forward-Backward Probability Dictionaries



```python
# Probability dictionaries for forward, backward, and forward-backward algos
initial_probs_fb = {
    "I": 0.2,
    "E": 0.8
}

transition_probs_fb = {
    "I": {"I": 0.1, "E": 0.9},
    "E": {"I": 0.6, "E": 0.4}
}

emission_probs_fb = {
    "I": {"A": 0.1, "T": 0.1, "C": 0.4, "G": 0.4},
    "E": {"A": 0.15, "T": 0.15, "C": 0.3, "G": 0.3}
}
```

## Demonstrate the StrMatrix Object
The StrMatrix object converts our nested probability dictionaries into log‑space NumPy matrices with string‑based indexing, allowing the HMM algorithms to access transition and emission values cleanly and consistently.

### Example StrMatrix Construction from a Basic Emission Probability Dictionary

```python
emit_probs1 = {
    "E": {"A": 0.3, "C": 0.2, "G": 0.2, "T": 0.3},
    "I": {"A": 0.1, "C": 0.4, "G": 0.4, "T": 0.1}
}

x = StrMatrix(emit_probs1)

display(show_matrix_html(x.get_matrix()))
```

### Demonstration of StrMatrix with Viterbi (I/G)

```python
# StrMatrix demonstration for Viterbi (I/G)
trans_v = StrMatrix(transition_probs_v)
emit_v = StrMatrix(emission_probs_v)

display(show_matrix_html(trans_v.get_matrix()))
display(show_matrix_html(emit_v.get_matrix()))
```

### Demonstration of StrMatrix with Forward/Backward (I/E)

```python
# StrMatrix demonstration for Forward–Backward (I/E)
trans_fb = StrMatrix(transition_probs_fb)
emit_fb = StrMatrix(emission_probs_fb)

display(show_matrix_html(trans_fb.get_matrix()))
display(show_matrix_html(emit_fb.get_matrix()))
```

## Instantiate Models
This cell creates the two HMM model instances used throughout the notebook: one configured with the I/G probabilities for the Viterbi algorithm, and one configured with the I/E probabilities for the Forward, Backward, and Forward–Backward algorithms.## Instantiate Models
This cell creates two separate HMM model instances—one for the Viterbi algorithm using the I/G probability set, and one for the Forward–Backward algorithms using the I/E probability set.

```python
# Instantiate Viterbi model (I/G)
model_v = HMMModel(initial_probs_v, transition_probs_v, emission_probs_v)

# Instantiate Forward–Backward model (I/E)
model_fb = HMMModel(initial_probs_fb, transition_probs_fb, emission_probs_fb)
```

## Define Observation Sequence
This cell defines the observation sequence that both HMM models will evaluate in all subsequent algorithm runs.

```python
# Defining observation sequence
obs = "ACGTGCTAGTCAGTACGATCGTAC"
```

## Execution Blocks
### Viterbi Algorithm
This cell runs the Viterbi algorithm on the observation sequence and displays the Viterbi dynamic‑programming matrix, the traceback matrix, and the final optimal hidden state path.

```python
# Run Viterbi algorithm and display all outputs
vmat, tmat, viterbi_path = model_v.viterbi_algorithm(obs)

display(show_matrix_html(vmat))
display(show_matrix_html(tmat))

viterbi_path
```

### Forward Algorithm
This cell runs the Forward algorithm on the observation sequence using the I/E model and outputs the forward probability matrix.

```python
# Run Forward algorithm and display forward matrix
fwd_matrix = model_fb.forward_algorithm(obs)

display(show_matrix_html(add_headers(fwd_matrix, model_fb.states, obs)))
fwd_matrix
```

### Backward Algorithm
This cell runs the Backward algorithm on the observation sequence using the I/E model and outputs the backward probability matrix.

```python
# Run Backward algorithm and display backward matrix
bwd_matrix = model_fb.backward_algorithm(obs)

display(show_matrix_html(add_headers(bwd_matrix, model_fb.states, obs)))
bwd_matrix
```

### Forward-Backward Algorithm
This cell runs the Forward–Backward algorithm on the observation sequence using the I/E model and outputs the MAP state sequence (computed from posterior probabilities) and the full forward–backward probability matrix.

```python
# Run Forward–Backward algorithm and display outputs
fb_states, fb_matrix = model_fb.forward_backward_algorithm(obs)

display(show_matrix_html(fb_matrix))
fb_states
```
