## Demonstrating Viterbi using a basic model

```python
obs1 = "ATGCAATAGCGCTAGATCGATCGATCGTAGCTAGCTAGCTAG"

init_probs1 = {
    "E": 0.6,
    "I": 0.4
}

trans_probs1 = {
    "E": {"E": 0.8, "I": 0.2},
    "I": {"E": 0.3, "I": 0.7}
}

emit_probs1 = {
    "E": {"A": 0.3, "C": 0.2, "G": 0.2, "T": 0.3},
    "I": {"A": 0.1, "C": 0.4, "G": 0.4, "T": 0.1}
}

model1 = HMMModel(initial_probs=init_probs1, transition_probs=trans_probs1, emission_probs=emit_probs1)
vmat1, tmat1, seq1 = model1.viterbi_algorithm(obs1)
print("".join(seq1), "\n")

print("=== TRACEBACK MATRIX ===")
display(show_matrix_html(tmat1))

print("=== VITERBI MATRIX ===")
display(show_matrix_html(vmat1))



```

## Demonstrating FWD

```python
obs2 = "GCGCGCATATATATAGCGCATATGCGCGCATATATATAGCGC"

init_probs2 = {
    "E": 0.7,
    "I": 0.3
}

trans_probs2 = {
    "E": {"E": 0.9, "I": 0.1},
    "I": {"E": 0.15, "I": 0.85}
}

emit_probs2 = {
    "E": {"A": 0.15, "C": 0.35, "G": 0.35, "T": 0.15},
    "I": {"A": 0.4, "C": 0.1, "G": 0.1, "T": 0.4}
}

model2 = HMMModel(initial_probs=init_probs2, transition_probs=trans_probs2, emission_probs=emit_probs2)
matrix2, prob2 = model2.forward_algorithm(obs2)
```
